```python
import pandas as pd
```

```python
data=pd.read_csv("C:/Users/lavanya/Downloads/Internship_Files/Raw_Dataset.csv")
df=pd.DataFrame(data)
print(df)
```

```python
df["Arrival_time"] = pd.to_timedelta(df["Arrival_time"])
df["Departure_Time"] = pd.to_timedelta(df["Departure_Time"])
```

```python
source = input("Enter source station: ").strip().upper()
destination = input("Enter destination station: ").strip().upper()
source_trains = df[df["Station_Name"].str.upper() == source]
destination_trains = df[df["Station_Name"].str.upper() == destination]
common_trains = set(source_trains["Train_No"]) & set(destination_trains["Train_No"])
results = []

for train in common_trains:
    train_data = df[df["Train_No"] == train]
    source_row = train_data[
        train_data["Station_Name"].str.upper() == source].iloc[0]
    destination_row = train_data[train_data["Station_Name"].str.upper() == destination].iloc[0]
    if source_row["SN"] < destination_row["SN"]:
        duration = destination_row["Arrival_time"] - source_row["Departure_Time"]

        if duration < pd.Timedelta(0):
            duration += pd.Timedelta("1 day")

results.append({"Train_No": train,"Source": source,"Destination": destination,"Departure_Time": str(source_row["Departure_Time"]).split()[-1],
"Arrival_Time": str(destination_row["Arrival_time"]).split()[-1],"Journey_Duration": str(duration).split()[-1]})

if results:
    result_df = pd.DataFrame(results)
    result_df = result_df.sort_values("Journey_Duration")
    print("\nAvailable Direct Trains:")
    print(result_df.to_string(index=False))
else:
    print("\nNo direct trains found for the selected route.")
```
