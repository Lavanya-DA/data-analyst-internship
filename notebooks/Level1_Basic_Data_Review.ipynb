```python
import pandas as pd
```

```python
data=pd.read_csv("C:/Users/lavanya/Downloads/Internship_Files/Raw_Dataset.csv")
df=pd.DataFrame(data)
print(df)
```

```python
print("Total Number of Rows:",df.shape[0])
print("Total Number of Columns:",df.shape[1])
print("Columns Name:",df.columns.tolist)
print("Basic Information of Dataset: \n")
df.info()
print("Statistical Summary of Numeric Columns:\n")
df.describe()
print("Columns Data Type :\n")
df.dtypes
```

```python
df.groupby("Train_No")["Station_Code"].agg(Starting_Station="first",Ending_Station="last")
```

```python
df.groupby("Train_No")["Station_Code"].agg(Number_of_Stops="count")
```

```python
Stops=df.groupby("Train_No")["Station_Code"].agg(Number_of_Stops="count")
Max_Stop=Stops["Number_of_Stops"].max()
Min_Stop=Stops["Number_of_Stops"].min()
print(Max_Stop)
print(Min_Stop)
```

```python
Train_With_Max_Stops=Stops[Stops["Number_of_Stops"]==Max_Stop]
```

```python
Train_With_Min_Stops=Stops[Stops["Number_of_Stops"]==Min_Stop]
```

```python
print("Maximum_Stop :\t\t",Max_Stop)
print("Train_With_Maximum_Stop \n\n",Train_With_Max_Stops)
print("\nMinimum_Stop :\t\t",Min_Stop)
print("Train_With_Minimum_Stops \n\n",Train_With_Min_Stops)
```
